# Timmy & Sarah think they are in love, but around where they live, they will only know once they pick a flower each. If one of the flowers has an even number of petals and the other has an odd number of petals it means they are in love.

# Write a function that will take the number of petals of each flower and return true if they are in love and false if they aren't.

[original link](https://www.codewars.com/kata/555086d53eac039a2a000083/train/python)

```python
def lovefunc( flower1, flower2 ):
    if flower1 % 2 == 0:
        flower1_petal = "Even"
    else:
        flower1_petal = "Odd"


    if flower2 % 2 == 0:
        flower2_petal = "Even"
    else:
        flower2_petal = "Odd"

    if (flower1_petal == "Odd" and flower2_petal == "Even") or \
       (flower1_petal == "Even" and flower2_petal == "Odd"):
            return True
    else:
        return False
```

```python
def run_love_tests():
    print("Running lovefunc tests...")

    # Test Case 1: Even and Odd (In love)
    assert lovefunc(4, 7) == True, "Test 1 Failed: 4 (Even) and 7 (Odd) should be True"

    # Test Case 2: Odd and Even (In love)
    assert lovefunc(23, 12) == True, "Test 2 Failed: 23 (Odd) and 12 (Even) should be True"

    # Test Case 3: Both Even (Not in love)
    assert lovefunc(2, 2) == False, "Test 3 Failed: 2 (Even) and 2 (Even) should be False"

    # Test Case 4: Both Odd (Not in love)
    assert lovefunc(5, 9) == False, "Test 4 Failed: 5 (Odd) and 9 (Odd) should be False"

    print("All flower tests passed successfully! 🌸 Your code is bulletproof.")

# Execute the tests
run_love_tests()
```
