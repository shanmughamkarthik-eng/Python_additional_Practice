1. The "Smart" Grocery List

Instead of just a list, build a program that uses functions to manage data.

The Goal: A script where you can add items, remove items, and view the total list.

```python
items_list = []

def add_item(item):
    items_list.append(item)

def remove_item(item):
    global items_list
    if item in items_list:
        items_list.remove(item)
        print("Item removed successfully!")
    else:
        print("Item not found in list.")

def view_list():
    print("\n--- Current List ---")
    for i in items_list: # Loop through to make it look nicer
        print(f"- {i}")
    print("--------------------\n")

print("The 'Smart' Grocery List")

while True:
    print("1. Add Item | 2. Remove Item | 3. View List | 4. Quit")
    case = int(input("Enter your choice: "))

    if case == 1:
        item = input("Enter the item: ")
        add_item(item)
    elif case == 2:
        item = input("Enter the item to remove: ")
        remove_item(item)
    elif case == 3:
        view_list()
    elif case == 4:
        print("Goodbye!")
        break
    else:
        print("Invalid choice!")
```
