#Project Requirements: BankAccount
##1. The Attributes (Data):
account_holder: The name of the person (String).
balance: The starting amount of money (Float/Int).

##2. The Methods (Actions):
###deposit(amount):
Should add the amount to the balance.
Security check: Ensure the amount is greater than 0. If it's negative, print an error.

###withdraw(amount):
Should subtract the amount from the balance.
Security check 1: Ensure the amount is greater than 0.
Security check 2: Ensure the user has enough money. If amount > balance, print "Insufficient funds!" and don't take any money.

###display_balance():
Should print a nice message showing the account holder's name and current balance.


###Your Goal:
Create the class.

Create an object for a user (e.g., "John" with 1000).

Try to deposit a negative number (it should fail).

Try to withdraw more than 1000 (it should fail).

Withdraw a valid amount and display the final balance.


```python
import os

class BankAccount:
  def __init__(self, account_holder, initial_balance):
    self.account_holder = account_holder
    self.filename = f"{self.account_holder}_balance.txt"
    self.__balance = self._load_balance(initial_balance)


# Method for load balance
  def _load_balance(self, default_balance):
    if os.path.exists(self.filename):
      with open(self.filename, "r") as f:
        return float(f.read())

    else:
      self._save_balance(default_balance)
      return default_balance

#Method for save balance
  def _save_balance(self, amount):
    with open(self.filename, "w") as f:
        f.write(str(amount))

#Method for deposit
  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
      self._save_balance(self.__balance)
      print(f"Deposited Rs.{amount}.\n Balance availabe: Rs.{self.__balance}")

    else:
      print("Invalid deposit amount. Amount must be greater than 0.")

#Methos for withdraw
  def withdraw(self, amount):
    if amount > 0:
      if amount <= self.__balance:
        self.__balance -= amount
        self._save_balance(self.__balance)
        print(f"Withdrew Rs.{amount}.\n Balance availabe: Rs.{self.__balance}")

      else:
        print("Insufficient funds!")

    else:
      print("Invalid withdrawal amount. Amount must be greater than 0.")

#Method for display balance
  def display_balance(self):
    print(f"Account Holder: {self.account_holder}\nBalance: Rs.{self.__balance}")

########################################
while True:
  try:
    name = input("Enter account holder's name: ")
    filename = f"{name}_balance.txt"

    if os.path.exists(filename):
       print(f"Welcome back, {name}!")
       user_account = BankAccount(name, 0)
       break
    else:
       print(f"Hello, {name}! you are a new customer.")

       balance = float(input("Enter initial deposit amount: "))
       if balance < 0:
        print("Initial deposit cannot be negative.")
        continue
       user_account = BankAccount(name, balance)
       break
  except ValueError:
    print("Invalid input. Please enter a valid amount (e.g., 500 or 500.50).")


while True:
  print("\n--- Bank Menu ---")
  print("1. Deposit")
  print("2. Withdraw")
  print("3. Check Balance")
  print("4. Exit")

  choice = input("\nEnter your choice: ")

  if choice == "1":
    try:
      amount = float(input("Enter deposit amount: "))
      user_account.deposit(amount)

    except ValueError:
      print("Invalid input. Please enter a valid amount (e.g., 500 or 500.50).")

  elif choice == "2":
    try:
      amount = float(input("Enter withdrawal amount: "))
      user_account.withdraw(amount)

    except ValueError:
      print("Invalid input. Please enter a valid amount (e.g., 500 or 500.50).")

  elif choice == "3":
    user_account.display_balance()

  elif choice == "4":
    print("Exiting the program.")
    break

  else:
    print("Invalid choice. Please select a valid option.")
```
