###**Dice Rolling Simulator**
This project helps you understand how computers handle randomness.
1. The Goal: Simulate rolling a six-sided die. The program should ask the user if they want to roll, generate a random number from 1 to 6, and display it. It should keep asking until the user decides to stop.

2. Skills Used: The random module, while loops, and basic conditional logic.

```python
import random
import time

print("Dice Rolling Simulator")

dice_faces = {
    1: " ------- \n|       |\n|   O   |\n|       |\n ------- ",
    2: " ------- \n| O     |\n|       |\n|     O |\n ------- ",
    3: " ------- \n| O     |\n|   O   |\n|     O |\n ------- ",
    4: " ------- \n| O   O |\n|       |\n| O   O |\n ------- ",
    5: " ------- \n| O   O |\n|   O   |\n| O   O |\n ------- ",
    6: " ------- \n| O   O |\n| O   O |\n| O   O |\n ------- "
}

while True:
    try:
        choice = input("\nDo you want to roll the dice(y/n?): ")

        if choice.lower() not in ['y','n']:
            raise ValueError

        elif choice.lower() == 'n':
            print("\n\tThanks for playing!")
            break

        else:
            number = random.randint(1,6)
            print("\nRolling...\n")
            time.sleep(2)
            print(dice_faces[number])
            # print(f"\n\tYour number is :{number}\n")

    except ValueError:
        print("\nInvalid input. Please enter 'y' or 'n'.\n")
```
