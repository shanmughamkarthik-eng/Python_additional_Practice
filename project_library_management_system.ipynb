###**The Project: Library Management System**

Build a system where you can add books, lend them to users, and track inventory

```python
class Book:
    def __init__(self, title, author, is_loaned):
        self.title = title
        self.author = author
        self.is_loaned = is_loaned

    def __str__(self):
        status = "Loaned" if self.is_loaned else "Available"
        return f"Title: {self.title}\nAuthor: {self.author}\nStatus: {status}"


class Library:
    def __init__(self):
        self.books = []

    def add_book(self):
        title = input("Enter book title: ")
        author = input("Enter book author: ")
        is_loaned = input("Is the book loaned? (yes/no): ")
        is_loaned = True if is_loaned.lower() == "yes" else False

        self.books.append(Book(title, author, is_loaned))

        print("\n\tBook added successfully!\n")


    def show_available_books(self):
        found = False
        for book in self.books:
            if not book.is_loaned:
                print(book)
                print("-" * 20)
                found = True

        if not found:
            print("\n\tNo available books found.\n")

    def borrow_book(self):
        title = input("Enter book title: ")
        for book in self.books:
            if book.title.lower() == title.lower() and not book.is_loaned:
                book.is_loaned = True
                print("\n\tBook borrowed successfully!\n")
                break

        else:
            print("\n\tBook not found or already loaned.\n")

    def return_book(self):
        title = input("Enter book title: ")
        for book in self.books:
            if book.title.lower() == title.lower() and book.is_loaned:
                book.is_loaned = False
                print("\n\tBook returned successfully!\n")
                break

        else:
            print("\n\tBook not found or not loaned.\n")

    def remove_book(self):
        title = input("Enter book title: ")
        for book in self.books:
            if book.title.lower() == title.lower():
                self.books.remove(book)
                print("\n\tBook removed successfully!\n")
                break

        else:
            print("\n\tBook not found.\n")

#---------------------------------------------------------------------

my_library = Library()

print("Library Management System")

while True:
    try:
        print("1. Add Book")
        print("2. Show Available Books")
        print("3. Borrow Book")
        print("4. Return Book")
        print("5. Remove Book")
        print("6. Exit")

        choice = int (input("Enter your choice: "))

        if choice == 1:
            print("\n\tAdd Book")
            my_library.add_book()

        elif choice == 2:
            print("\n\tShow Available Books\n")
            my_library.show_available_books()

        elif choice == 3:
            print("\n\tBorrow Book\n")
            my_library.borrow_book()

        elif choice == 4:
            print("\n\tReturn Book\n")
            my_library.return_book()

        elif choice == 5:
            my_library.remove_book()

        elif choice == 6:
            print("\n\tExiting...")
            break

        else:
            print("\n\tInvalid choice. Please try again.\n")

    except ValueError:
        print("Invalid input. Please enter a number.")
        continue
```
