Your classmates asked you to copy some paperwork for them. You know that there are 'n' classmates and the paperwork has 'm' pages.

Your task is to calculate how many blank pages do you need.

If n < 0 or m < 0 return 0.

**Example:**


> n= 5, m=5: 25

> n=-5, m=5:  0










```python
def paperwork(n, m):
    # Happy Coding! ^_^
    if (n < 0 or m < 0):
        return 0

    else:
        return n * m
```

```python
def run_paperwork_tests():
    print("Running paperwork tests...")

    # Test Case 1: Standard positive numbers (5 classmates * 5 pages)
    assert paperwork(5, 5) == 25, "Test 1 Failed: 5 * 5 should be 25"

    # Test Case 2: Negative classmates (Invalid input)
    assert paperwork(-5, 5) == 0, "Test 2 Failed: Negative 'n' should return 0"

    # Test Case 3: Negative pages (Invalid input)
    assert paperwork(5, -5) == 0, "Test 3 Failed: Negative 'm' should return 0"

    # Test Case 4: Zero check (0 classmates means 0 pages)
    assert paperwork(0, 10) == 0, "Test 4 Failed: 0 classmates should return 0"

    print("All paperwork tests passed flawlessly! 📝 Excellent job.")

# Execute the tests
run_paperwork_tests()
```
