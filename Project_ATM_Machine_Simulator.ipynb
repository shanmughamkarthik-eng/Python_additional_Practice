####ATM Machine Simulator
#####Create an ATM machine simulator with the following features:

1. Initial balance: 10000
2. PIN: 1234 (3 attempts allowed)
3. Menu options:
  *   Check Balance
  *   Withdraw Money (check for sufficient balance and valid amount)
  *   Deposit Money
  *   Change PIN
  *   Exit
4. Keep transaction history (last 5 transactions)
5. Input validation for all operations

Hint: Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

```python
def withdraw_money():
  global balance
  global transaction_history

  while True:
    try:
      amount = float(input("Enter the amount: "))
    except ValueError:
      print("Please enter a numeric value.")
      continue

    if amount <= 0:
      print("Invalid amount. Please enter a positive number.")
      continue

    elif amount > balance:
      print("Insufficient balance.")
      continue

    else:
      balance -= amount
      transaction_history.append(f" - {amount}")
      print(f"Withdrawal successful. New balance: {balance}")
      return

#----------------------------------------------------------------

def deposit_money():
  global balance
  global transaction_history

  while True:
    try:
      amount = float(input("Enter the amount: "))
    except ValueError:
      print("Please enter a numeric value.")
      continue

    if amount <= 0:
      print("Invalid amount. Please enter a positive number.")
      continue

    else:
      balance += amount
      transaction_history.append(f" + {amount}")
      print(f"Deposit successful. New balance: {balance}")
      return

#-----------------------------------------------------------------

def check_balance():
  global balance
  global transaction_history

  print(f"Your balance is: {balance:,.2f}")

  if input("Do you want to see the last 5 transactions? (y/n): ").lower() == 'y':
     # This slices the list to get the last 5 items
    last_five = transaction_history[-5:]

    if not last_five:
      print("No transactions yet.")

    else:
      print("\n--- Recent Transactions ---")
      for tx in last_five:
        print(tx)
      return

  else:
    return

#-------------------------------------------------------------------

def change_pin():
  global pin
  global transaction_history
  attempts = 0

  while attempts < 3:
    try:
      current = int(input(f"Enter your current PIN (Attempt {attempts + 1}/3):"))

      if current != pin:
          print("Incorrect PIN. Authorization failed.")
          attempts += 1
          continue

      else:
        try:
          new_pin = int(input("Enter the new PIN: "))
          confirmation = int(input("Confirm the new PIN: "))

          if new_pin == confirmation:
            pin = new_pin
            print("PIN changed successfully.")
            return True

          else:
            print("Pin not matched")
            return False

        except ValueError:
          print("Please enter a numeric value.")
          continue

    except ValueError:
      print("Please enter a numeric value.")
      attempts += 1
      continue

  if attempts >= 3:
    print("Maximum number of attempts reached. Exiting.")
    return False

#-------------------------------------------------------------------------
def login():
  global pin
  attempts = 0

  while attempts < 3:
    try:
      entered_pin = int(input(f"Enter your PIN  (Attempt {attempts + 1}/3): "))
      if entered_pin == pin:
        print("Login successful.")
        return True

      else:
          print("Incorrect PIN. Authorization failed.")
          attempts += 1
          continue

    except ValueError:
      print("Please enter a numeric value.")
      attempts += 1
      continue

  return False

#-----------------------------------------------

print("ATM Machine Simulator")
balance = 10000
pin = 1234
transaction_history = []

if not login():
  print("Exiting")

else:
  while True:
    print("\n1. Check Balance")
    print("2. Withdraw Money")
    print("3. Deposit Money")
    print("4. Change PIN")
    print("5. Exit")

    try:
      choice = int(input("\nEnter your choice: "))
    except ValueError:
      print("Invalid input. Please enter a valid number.")
      continue

    if choice == 5:
      print("Exiting")
      break

    elif choice == 1:
      check_balance()

    elif choice == 2:
      withdraw_money()

    elif choice == 3:
      deposit_money()

    elif choice == 4:
      if not change_pin():
        break

    else: print("Invalid choice. Please select a valid option. eg., 1 to 5")
```
